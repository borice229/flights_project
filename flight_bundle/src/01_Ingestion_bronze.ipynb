```python
catalog = dbutils.widgets.get("catalog");
spark.sql(f"USE CATALOG {catalog}");
spark.sql("CREATE SCHEMA IF NOT EXISTS bronze")
```

```python
path_airport = "/Volumes/samples/databricks/datasets/flights/airport-codes-na.txt"
path_depart ="/Volumes/samples/databricks/datasets/flights/departuredelays.csv"
```

**Airport dataset**

```python
df_airport = spark.read.csv(path_airport, header=True, inferSchema=True, sep='\t')
display(df_airport.limit(5))
```

```python
df_airport.write.format('delta').mode('overwrite').saveAsTable('bronze.airport_raws')
```

Departement dataset

```python
df_depart = spark.read.csv(path_depart, header=True, inferSchema=True)
display(df_depart.limit(5))
```

```python
df_depart.write.format('delta').mode('overwrite').saveAsTable("bronze.depart_raws")
```
